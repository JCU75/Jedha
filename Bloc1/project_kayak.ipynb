### 1.1.1 Get GPS data for city list

```python
def get_coordinates(city_name):
    """
    Return latitude and longitude of the city
    """

    URL_NOMINATIM = "https://nominatim.openstreetmap.org/search"
    params = {
    "q": city_name,
    "format": "json"
    }
    headers = {
        'User-Agent': 'MyPythonScript/1.0',
        'Referer': 'https://mywebsite.com/page.html'
    }          


    try:
        response = requests.get(URL_NOMINATIM, params=params, headers=headers)
        response.raise_for_status()  # Raise an exception for non-200 status codes

        data = response.json()

        if len(data) == 0:
            print("City not found")
            return None # Return None to indicate unsuccessful retrieval
       
        latitude = data[0]['lat']
        longitude = data[0]['lon']
        return [latitude, longitude]
  
    except requests.exceptions.RequestException as e:
        print(f"Failed to retrieve data: {e}")
        return None # Return None to indicate unsuccessful retrieval

```

```python
city_data = []
for city in city_list:
    city_coordinates = get_coordinates(city)
    current_city = {'city_name': city, "coordinates": city_coordinates}
    #print(current_city)
    city_data.append(current_city)
    time.sleep(1) # Timer to avoid overloading the API 

print(city_data)    
```

```python
def get_forecast_data(lat,lon):
    """
    This function takes a city latitude and a city longitude as input and returns a forecast data for 5 days.
    """
    exclude = "hourly,minutely"
    url_forecast_5days = f"https://api.openweathermap.org/data/2.5/forecast?lat={lat}&lon={lon}&exclude={exclude}&appid={API_KEY_WEATHER}&units=metric"
    
    try:
        response = requests.get(url_forecast_5days)
        response.raise_for_status()  # Raise an exception for non-200 status codes

        data = response.json()

        if len(data) == 0:
            print("City not found")
            return None # Return None to indicate unsuccessful retrieval
       
        return data
  
    except requests.exceptions.RequestException as e:
        print(f"Failed to retrieve data: {e}")
        return None # Return None to indicate unsuccessful retrieval
   
```

```python
weather_data = df_weather_data.to_dict(orient='records')
display(weather_data)
```

```python
for city in weather_data:
    print(city)
```

```python
# get the weather data for each city in the list
city_weather_data = []
for city in weather_data:
    city_name = city['city_name']
    lat = city['latitude']
    lon = city['longitude']
    city_weather = get_forecast_data(lat, lon)
    #print(city_weather)
    current_city = {"city": city_name , "weather_info" : city_weather}
    city_weather_data.append(current_city)
    time.sleep(0.2)

display(city_weather_data) 
```

```python
forecast_filtered = []
for city in city_weather_data:
    city_name  = city["city"]
    #print(city_name)
    weather_info = city["weather_info"]
    weather_info_data = weather_info["list"]
    for info in weather_info_data:
        day = info["dt_txt"].split(" ")[0]
        #print(day)
        temp = info["main"]["temp"]
        #print(temp)
        weather = info["weather"][0]["description"]
        #print(weather)
        wind_speed = info["wind"]["speed"]
        #print(wind_speed)
        humidity = info["main"]["humidity"]
        #print(humidity)
        row = {"city_name":city_name, "day": day, "temp": temp, "weather": weather, "wind_speed": wind_speed, "humidity": humidity}
        forecast_filtered.append(row)

display(forecast_filtered)        
       
    
```

```python
df = pd.DataFrame(forecast_filtered)
df.head()
```

```python
# Convert 'day' column to datetime type to make grouping easier
df['day'] = pd.to_datetime(df['day'])
df.head()
```

```python
# Grouping by day and calculating statistics
grouped_df = df.groupby(['city_name', 'day']).agg({
    'temp': 'mean',
    'wind_speed': 'mean',
    'humidity': 'mean',
    'weather': lambda x: x.mode()[0]
}).reset_index()

grouped_df
```

```python
df_city_data = pd.merge(df_weather_data, grouped_df, on='city_name')

display(df_city_data)
```

```python
# Create mapping icon to display on the map
icon_map = {
    "clear sky": "https://img.icons8.com/emoji/96/sun-emoji.png",
    "few clouds": "https://img.icons8.com/emoji/96/cloud-emoji.png",
    "scattered clouds": "https://img.icons8.com/emoji/96/cloud-emoji.png",
    "broken clouds": "https://img.icons8.com/fluency/96/clouds--v3.png",
    "overcast clouds": "https://img.icons8.com/fluency/96/clouds--v3.png",
    "shower rain": "https://img.icons8.com/emoji/96/cloud-with-rain-emoji.png",
    "light rain": "https://img.icons8.com/emoji/96/cloud-with-rain-emoji.png",
    "heavy intensity rain": "https://img.icons8.com/color/96/intense-rain--v1.png",
    "rain": "https://img.icons8.com/color/96/intense-rain--v1.png",
    "thunderstorm": "https://img.icons8.com/emoji/96/cloud-with-lightning-and-rain.png",
    "snow": "https://img.icons8.com/emoji/96/snowflake-emoji.png",
    "mist": "https://img.icons8.com/emoji/96/fog.png"
}
```

```python
# Create new column with mapping icon
df_forecast['icon_url'] = df_forecast['weather'].map(icon_map).fillna("https://img.icons8.com/ios-glyphs/30/question-mark.png")
```

```python
def bbox_for_point(lon, lat, km=25):
    """Return mapbox coordinates for image box.
       Mapbox expects order: Top-Left, Top-Right, Bottom-Right, Bottom-Left"""
    lat_deg = km / 110.574
    lon_deg = km / (111.320 * cos(radians(lat)))
    return [
        [lon - lon_deg, lat + lat_deg],  # Top-Left
        [lon + lon_deg, lat + lat_deg],  # Top-Right
        [lon + lon_deg, lat - lat_deg],  # Bottom-Right
        [lon - lon_deg, lat - lat_deg],  # Bottom-Left
    ]

def make_layers(df_subset, km=25):
    layers = []
    for _, r in df_subset.iterrows():
        layers.append({
            "sourcetype": "image",
            "source": r["icon_url"],
            "coordinates": bbox_for_point(r["longitude"], r["latitude"], km=km)
        })
    return layers
```

```python
## Make figure & frames

days = sorted(df_forecast['day'].unique())
if len(days) == 0:
    raise ValueError("Aucune date trouvée dans df_forecast['day']")

# Choose non-empty day0 for initial display
first_day = next((d for d in days if not df_forecast[df_forecast['day']==d].empty), days[0])
df_forecast0 = df_forecast[df_forecast['day'] == first_day]

fig = go.Figure()

# "invisible" trace needed to display the map and provide hover points
fig.add_trace(go.Scattermapbox(
    lat=df_forecast0["latitude"],
    lon=df_forecast0["longitude"],
    mode="markers",
    # invisible but hoverable markers: size not zero, color transparent
    marker=dict(size=18, color="rgba(0,0,0,0)"),
    text=df_forecast0["city_name"],
    hovertemplate="%{text}<extra></extra>",
    showlegend=False
))

# Layout initial: 
fig.update_layout(
    mapbox_style="carto-darkmatter",
    mapbox_layers=make_layers(df_forecast0, km=25),
    mapbox_center={"lat": 46.6, "lon": 2.2},
    mapbox_zoom=5,
    width=1200, height=800,
    title="Météo (icônes) — animation"
)

# Build frames : each frame contains the trace of hover + layout (layers + center + zoom)
frames = []
for day in days:
    d = df_forecast[df_forecast['day'] == day]
    # trace : same points (invisible) but with text = city_name for the hover
    trace = go.Scattermapbox(
        lat=d["latitude"],
        lon=d["longitude"],
        mode="markers",
        marker=dict(size=36, color="rgba(0,0,0,0)"),
        text=d["city_name"],
        hovertemplate="%{text}<extra></extra>",
        showlegend=False
    )
    frame_layout = go.Layout(mapbox=dict(layers=make_layers(d, km=25)))
    frames.append(go.Frame(name=day, data=[trace], layout=frame_layout))

fig.frames = frames

# controls
fig.update_layout(
    updatemenus=[{
        "type": "buttons",
        "buttons": [{
            "label": "▶ Play",
            "method": "animate",
            "args": [None, {"frame": {"duration": 800, "redraw": True}, "fromcurrent": True}]
        }]
    }],
    sliders=[{
        "active": 0,
        "currentvalue": {"prefix": "Jour : "},
        "pad": {"t": 40},
        "steps": [{
            "args": [[d], {"frame": {"duration": 0, "redraw": True}, "mode": "immediate"}],
            "label": d,
            "method": "animate"
        } for d in days]
    }]
)

fig.show()
fig.write_html("top-5-city.html", auto_open=True)
```
